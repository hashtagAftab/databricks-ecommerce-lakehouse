```python
users = spark.read.table(f'{TABLE_PREFIX}.silver_users')
```

```python
%sql
create table if not exists projects.ecommerce.dim_users (
    user_sk bigint generated always as identity,
    user_id int,
    first_name string,
    last_name string,
    maiden_name string,
    gender string,
    email string,
    phone string,
    username string,
    birth_date date,
    image string,
    blood_group string,
    height float,
    weight float,
    eye_color string,
    hair_color string,
    hair_type string,
    street_address string,
    city string,
    state string,
    state_code string,
    postal_code string,
    lat_code float,
    long_code float,
    country string,
    university string,
    card_expire date,
    card_type string,
    currency string,
    company_dept string,
    company_name string,
    job_title string,
    office_street_address string,
    office_city string,
    office_state string,
    office_state_code string,
    office_postal_code string,
    office_lat float,
    office_long float,
    office_country string,
    user_agent string,
    crypto_coin string,
    crypto_network string,
    role string,
    effective_date date,
    expiration_date date default date '9999-12-31',
    is_current boolean default true
)
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported');
```

```python
%sql
merge into projects.ecommerce.dim_users t
using projects.ecommerce.silver_users s
on t.user_id = s.user_id and t.is_current = true
when matched and (
    t.first_name <> s.first_name or
    t.last_name <> s.last_name or
    t.maiden_name <> s.maiden_name or
    t.birth_date <> s.birth_date or
    t.street_address <> s.street_address or
    t.city <> s.city or
    t.state <> s.state or
    t.state_code <> s.state_code or
    t.postal_code <> s.postal_code or
    t.country <> s.country or
    t.company_dept <> s.company_dept or
    t.company_name <> s.company_name or
    t.job_title <> s.job_title or
    t.office_street_address <> s.office_street_address or
    t.office_city <> s.office_city or
    t.office_state <> s.office_state or
    t.office_postal_code <> s.office_postal_code or
    t.office_country <> s.office_country
)
then
update set
    expiration_date = date_sub(current_date(), 1),
    is_current = false

when matched and (
    t.gender <> s.gender or
    t.email <> s.email or
    t.phone <> s.phone or
    t.username <> s.username or
    t.image <> s.image or
    t.blood_group <> s.blood_group or
    t.height <> s.height or
    t.weight <> s.weight or
    t.eye_color <> s.eye_color or
    t.hair_color <> s.hair_color or
    t.hair_type <> s.hair_type or
    t.lat_code <> s.lat_code or
    t.long_code <> s.long_code or
    t.university <> s.university or
    t.card_expire <> s.card_expire or
    t.card_type <> s.card_type or
    t.currency <> s.currency or
    t.office_state_code <> s.office_state_code or
    t.office_lat <> s.office_lat or
    t.office_long <> s.office_long or
    t.user_agent <> s.user_agent or
    t.crypto_coin <> s.crypto_coin or
    t.crypto_network <> s.crypto_network or
    t.role <> s.role
)
then
update set
    gender = s.gender,
    email = s.email,
    phone = s.phone,
    username = s.username,
    image = s.image,
    blood_group = s.blood_group,
    height = s.height,
    weight = s.weight,
    eye_color = s.eye_color,
    hair_color = s.hair_color,
    hair_type = s.hair_type,
    lat_code = s.lat_code,
    long_code = s.long_code,
    university = s.university,
    card_expire = s.card_expire,
    card_type = s.card_type,
    currency = s.currency,
    office_state_code = s.office_state_code,
    office_lat = s.office_lat,
    office_long = s.office_long,
    user_agent = s.user_agent,
    crypto_coin = s.crypto_coin,
    crypto_network = s.crypto_network,
    role = s.role

when not matched
then
insert (
    user_id,
    first_name,
    last_name,
    maiden_name,
    gender,
    email,
    phone,
    username,
    birth_date,
    image,
    blood_group,
    height,
    weight,
    eye_color,
    hair_color,
    hair_type,
    street_address,
    city,
    state,
    state_code,
    postal_code,
    lat_code,
    long_code,
    country,
    university,
    card_expire,
    card_type,
    currency,
    company_dept,
    company_name,
    job_title,
    office_street_address,
    office_city,
    office_state,
    office_state_code,
    office_postal_code,
    office_lat,
    office_long,
    office_country,
    user_agent,
    crypto_coin,
    crypto_network,
    role,
    effective_date
)
values (
    s.user_id,
    s.first_name,
    s.last_name,
    s.maiden_name,
    s.gender,
    s.email,
    s.phone,
    s.username,
    s.birth_date,
    s.image,
    s.blood_group,
    s.height,
    s.weight,
    s.eye_color,
    s.hair_color,
    s.hair_type,
    s.street_address,
    s.city,
    s.state,
    s.state_code,
    s.postal_code,
    s.lat_code,
    s.long_code,
    s.country,
    s.university,
    s.card_expire,
    s.card_type,
    s.currency,
    s.company_dept,
    s.company_name,
    s.job_title,
    s.office_street_address,
    s.office_city,
    s.office_state,
    s.office_state_code,
    s.office_postal_code,
    s.office_lat,
    s.office_long,
    s.office_country,
    s.user_agent,
    s.crypto_coin,
    s.crypto_network,
    s.role,
    current_date()
);
```

```python
%sql
insert into projects.ecommerce.dim_users (
    user_id,
    first_name,
    last_name,
    maiden_name,
    gender,
    email,
    phone,
    username,
    birth_date,
    image,
    blood_group,
    height,
    weight,
    eye_color,
    hair_color,
    hair_type,
    street_address,
    city,
    state,
    state_code,
    postal_code,
    lat_code,
    long_code,
    country,
    university,
    card_expire,
    card_type,
    currency,
    company_dept,
    company_name,
    job_title,
    office_street_address,
    office_city,
    office_state,
    office_state_code,
    office_postal_code,
    office_lat,
    office_long,
    office_country,
    user_agent,
    crypto_coin,
    crypto_network,
    role,
    effective_date
)
select 
    s.user_id,
    s.first_name,
    s.last_name,
    s.maiden_name,
    s.gender,
    s.email,
    s.phone,
    s.username,
    s.birth_date,
    s.image,
    s.blood_group,
    s.height,
    s.weight,
    s.eye_color,
    s.hair_color,
    s.hair_type,
    s.street_address,
    s.city,
    s.state,
    s.state_code,
    s.postal_code,
    s.lat_code,
    s.long_code,
    s.country,
    s.university,
    s.card_expire,
    s.card_type,
    s.currency,
    s.company_dept,
    s.company_name,
    s.job_title,
    s.office_street_address,
    s.office_city,
    s.office_state,
    s.office_state_code,
    s.office_postal_code,
    s.office_lat,
    s.office_long,
    s.office_country,
    s.user_agent,
    s.crypto_coin,
    s.crypto_network,
    s.role,
    current_date() as effective_date
from projects.ecommerce.silver_users s
inner join projects.ecommerce.dim_users t
    on t.user_id = s.user_id
where t.expiration_date = date_sub(current_date(), 1)
    and t.is_current = false;
```
