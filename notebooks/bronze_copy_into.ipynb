```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
from src.config import *
from src.schemas import *
```

```python
for endpoint in VALID_ENDPOINTS:
    table_name = f'{TABLE_PREFIX}.bronze_{endpoint}'
    schema = ENDPOINT_SCHEMAS[endpoint]
    
    df = spark.createDataFrame([], schema=schema)\
        .withColumn('ingestion_timestamp', F.current_timestamp())\
        .withColumn('source_system', F.lit(f'{BASE_URL}/{endpoint}'))
    
    if not spark.catalog.tableExists(table_name):
        df.writeTo(table_name).create()
        print(f'Created {table_name}')
    else:
        print(f'{table_name} already exists, skipping creation')
```
